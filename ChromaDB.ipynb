**Create ChromaDB client**

```python
import chromadb

client = chromadb.Client()
```

**Create ChromaDB collection**

```python
collection = client.get_or_create_collection(name="my_collections")
```

**Add data to ChromaDB collection**

```python
cricket_news = "India defeated Ireland in the T20 World Cup with Jasprit Bumrah delivering a match-winning performance."

football_news = "Real Madrid secured a place in the UEFA Champions League final after a thrilling victory over Manchester City."

election_news = "The presidential election campaign intensified as candidates focused on the economy, education, and healthcare."

ai_revolution_news = "AI-powered healthcare systems are improving disease detection and patient care worldwide."
```

```python
collection.add(
    documents = [cricket_news, football_news, election_news, ai_revolution_news],
    metadatas = [{"source": "cricket"},{"source": "football"},{'source':'election'},{"source":"ai revolution"}],
    ids = ["id1", "id2", "id3", "id4"]
)
```

**Similarity search**

```python
results = collection.query(
    query_texts=["technology"],
    n_results=2
)

results


collection.count()
```
